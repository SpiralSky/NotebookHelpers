## Export

Run the sync cell, restore `* 2`, rerun calculations, and save this notebook.
Setup/status cells are tagged `load-clean-remove`. Import links export ordinary
imports and never synchronize files; copy/skeleton requests export generated
Python. The helper must be importable in the exported notebook’s environment.

From the repository root, with the project installed, run on Linux/macOS:

```sh
load-clean check examples/dev_sync.ipynb --source-root examples
PYTHONPATH="$PWD/examples" load-clean build examples/dev_sync.ipynb --source-root examples --execute
```

This creates `build/sync.ipynb`; `42` is the calculation output. On other systems,
configure `PYTHONPATH` to the absolute `examples/` directory before building.
Review helper changes with Git before keeping them. Build/check never write
changes back to `sync_helpers.py`.
